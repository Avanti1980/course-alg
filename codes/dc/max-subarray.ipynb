分治

```python
def find_max_subarray(a: list, low: int, high: int) -> tuple[int, int, int]:
    # 返回最大子数组的起始索引、结束索引、和
    if low == high:
        return low, high, a[low]

    mid = (low + high) // 2

    # 递归求解左半数组
    l_low, l_high, l_sum = find_max_subarray(a, low, mid)

    # 递归求解右半数组
    r_low, r_high, r_sum = find_max_subarray(a, mid + 1, high)

    # 处理跨越中点的情况
    c_low, c_high, c_sum = find_max_cross_subarray(a, low, mid, high)

    # 三种情况取最大
    return max(
        (l_low, l_high, l_sum),
        (r_low, r_high, r_sum),
        (c_low, c_high, c_sum),
        key=lambda x: x[2],
    )


def find_max_cross_subarray(a: list, low: int, mid: int, high: int) -> tuple[int, int, int]:
    l_sum = r_sum = float("-inf")

    s = 0  # 找结束于 mid 的最大子数组的起始索引
    for i in range(mid, low-1, -1):  # 遍历 mid -> low
        s += a[i]
        if s > l_sum:
            l_sum, l_index = s, i

    s = 0  # 找开始于 mid+1 的最大子数组的结束索引
    for i in range(mid+1, high+1):  # 遍历 mid+1 -> high
        s += a[i]
        if s > r_sum:
            r_sum, r_index = s, i

    return l_index, r_index, l_sum + r_sum


a = [13, -3, -25, 20, -3, -16, -23, 18, 20, -7, 12, -5, -22, 15, -4, 7]
n = len(a)
print(find_max_subarray(a, 0, n - 1))
```

动态规划

```python
def find_max_subarray_dp(a: list) -> tuple[int, int, float]:
    n = len(a)
    d = [0] * n  # d[i]为以a[i]作结尾的最大子数组的和
    s = [0] * n  # s[i]为以a[i]作结尾的最大子数组的起始索引
    d[0], s[0] = a[0], 0
    for i in range(1, n):
        if d[i - 1] >= 0:
            d[i] = d[i - 1] + a[i]  # a[i]接在a[j,...,i-1]后面
            s[i] = s[i - 1]         # 继承其起始索引
        else:
            d[i] = a[i]          # a[i]不接在a[j,...,i-1]后面 另起炉灶
            s[i] = i             # 起始索引就是当前位置
    max_index = d.index(max(d))  # 遍历d获取最大元的索引
    return s[max_index], max_index, d[max_index]


find_max_subarray_dp(a)
```

d[i]只依赖d[i-1]，因此不需要维护完整的d表，可节省存储空间，s表也不需要，可在循环中同步维护全局最优

```python
def find_max_subarray_dp2(a: list) -> tuple[int, int, float]:
    n = len(a)
    cur_sum = a[0]          # 以当前元素结尾的最大子数组和
    cur_start = 0           # 该段的起始下标
    best_sum = cur_sum      # 全局最优和
    best_start = 0          # 全局最优段起始
    best_end = 0            # 全局最优段结束
    for i in range(1, n):
        if cur_sum >= 0:
            cur_sum += a[i]  # a[i]接在a[j,...,i-1]后面
        else:
            cur_sum = a[i]   # a[i]不接在a[j,...,i-1]后面 另起炉灶
            cur_start = i    # 起始索引就是当前位置
        if cur_sum > best_sum:
            best_sum = cur_sum
            best_start = cur_start
            best_end = i
    return best_start, best_end, best_sum


find_max_subarray_dp2(a)
```
