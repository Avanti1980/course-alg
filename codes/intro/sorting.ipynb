冒泡排序

```python
from typing import List


def bubble_sort(a: List[int]) -> None:
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1, i, -1):  # 从后往前遍历，将较小的元素像气泡一样浮到前面
            if a[j] < a[j - 1]:
                a[j], a[j - 1] = a[j - 1], a[j]


a = [5, 2, 4, 6, 1, 3]
bubble_sort(a)
a
```

引入swapped变量检测外层循环某一轮是否有翻转发生，如果没有则当前数组已排好序，可将最好情况下运行时间改善到$\Theta(n)$

```python
def bubble_sort_opt(a: List[int]) -> None:
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1, i, -1):  # 从后往前遍历，将较小的元素像气泡一样浮到前面
            if a[j] < a[j - 1]:
                a[j], a[j - 1] = a[j - 1], a[j]
                swapped = True
        if not swapped:
            break


a = [5, 2, 4, 6, 1, 3]
bubble_sort_opt(a)
a
```

选择排序

```python
def selection_sort(a: List[int]) -> None:
    n = len(a)
    for i in range(n - 1):
        smallest = i
        for j in range(i + 1, n):
            if a[j] < a[smallest]:
                smallest = j
        if smallest != i:  # 只有当最小元素不是当前元素时才交换
            a[i], a[smallest] = a[smallest], a[i]

a = [5, 2, 4, 6, 1, 3]
selection_sort(a)
a
```

插入排序

```python
def insertion_sort(a: List[int]) -> None:
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and key < a[j]:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key


a = [5, 2, 4, 6, 1, 3]
insertion_sort(a)
a
```

归并排序

```python
def merge(a: List[int], low: int, mid: int, high: int) -> None:
    # 拷贝数据到临时数组left
    left = a[low:mid + 1]
    # 拷贝数据到临时数组right
    right = a[mid + 1:high + 1]

    i, j, k = 0, 0, low

    # 归并left和right到a[low,...,high]
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            a[k] = left[i]
            i += 1
        else:
            a[k] = right[j]
            j += 1
        k += 1

    if i < len(left):  # 拷贝left的剩余元素
        a[k:high + 1] = left[i:]
    else:  # 拷贝right的剩余元素
        a[k:high + 1] = right[j:]


def merge_sort(a: List[int], low: int = 0, high: int = None) -> None:
    if high is None:
        high = len(a) - 1
    if low < high:
        mid = (low + high) // 2  # 中间点
        merge_sort(a, low, mid)
        merge_sort(a, mid + 1, high)
        merge(a, low, mid, high)


a = [5, 2, 4, 6, 1, 3]
merge_sort(a)
a
```

快速排序

```python
def partition(a: List[int], low: int, high: int) -> int:
    pivot = a[high]  # 最右元素作为主元
    i = low  # i指向小于主元的元素的存放位置 初始为最左

    for j in range(low, high):  # low -> high-1 遍历其它元素
        if a[j] <= pivot:
            a[i], a[j] = a[j], a[i]  # 小于主元的元素放到主元左边
            i += 1  # 存放位置右移一位

    # 所有小于主元的元素已位于主元左边
    # 当前的i就是主元应该放的位置
    # 当前的a[i]大于主元
    a[i], a[high] = a[high], a[i]

    return i


def quick_sort(a: List[int], low: int = 0, high: int = None) -> None:
    if high is None:
        high = len(a) - 1
    if low < high:
        m = partition(a, low, high)
        quick_sort(a, low, m - 1)
        quick_sort(a, m + 1, high)


a = [5, 2, 4, 6, 1, 3]
quick_sort(a)
a
```
